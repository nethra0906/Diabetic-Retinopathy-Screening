```python
from pathlib import Path
import numpy as np

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

INTERNAL_TEST_DIR = (
    PROJECT_ROOT
    / "results"
    / "improvement"
    / "internal_test"
)

CHUNK_DIR = INTERNAL_TEST_DIR / "tta_chunks"

# Your actual structure: files are directly inside tta_chunks/
prob_files = sorted(
    CHUNK_DIR.glob("probs_*.npy")
)

label_files = sorted(
    CHUNK_DIR.glob("labels_*.npy")
)

print("Probability chunks found:", len(prob_files))
print("Label chunks found:", len(label_files))

print("\nFirst probability:")
print(prob_files[0] if prob_files else "NONE")

print("\nLast probability:")
print(prob_files[-1] if prob_files else "NONE")

print("\nFirst label:")
print(label_files[0] if label_files else "NONE")

print("\nLast label:")
print(label_files[-1] if label_files else "NONE")
```

```python
print("Probability chunks:")

for i, p in enumerate(prob_files):
    print(i + 1, p.name)

print("\nLabel chunks:")

for i, p in enumerate(label_files):
    print(i + 1, p.name)
```

```python
print("Checking chunk sizes...\n")

for p, l in zip(prob_files, label_files):

    probs = np.load(p)
    labels = np.load(l)

    print(
        f"{p.name:25s}"
        f" probs={probs.shape}"
        f" labels={labels.shape}"
    )
```

```python
prob_chunks = [
    np.load(p)
    for p in prob_files
]

label_chunks = [
    np.load(p)
    for p in label_files
]

test_tta_probs = np.concatenate(
    prob_chunks,
    axis=0
)

test_tta_y_true = np.concatenate(
    label_chunks,
    axis=0
)

print("TTA probabilities:")
print(test_tta_probs.shape)

print("\nTest labels:")
print(test_tta_y_true.shape)
```

```python
del prob_chunks
del label_chunks

print("✅ Chunk arrays released")
```

```python
test_tta_probs = test_tta_probs.astype(
    np.float64
)

test_tta_probs = (
    test_tta_probs
    / np.clip(
        test_tta_probs.sum(
            axis=1,
            keepdims=True
        ),
        1e-12,
        None
    )
)

print(
    "First 5 probability row sums:"
)

print(
    test_tta_probs[:5].sum(axis=1)
)
```

```python
test_tta_predictions = np.argmax(
    test_tta_probs,
    axis=1
)

print(
    "Prediction shape:",
    test_tta_predictions.shape
)

print(
    "Predicted classes:",
    np.unique(test_tta_predictions)
)
```

```python
threshold_file = (
    INTERNAL_TEST_DIR
    / "frozen_thresholds.json"
)

print(
    "Threshold file exists:",
    threshold_file.exists()
)
```

```python
import json

with open(
    threshold_file,
    "r",
    encoding="utf-8"
) as f:

    threshold_info = json.load(f)

print(threshold_info)

best_thresholds = np.asarray(
    threshold_info["thresholds"],
    dtype=np.float64
)

print(
    "\nFrozen thresholds:",
    best_thresholds
)
```

```python
severity_values = np.arange(
    5,
    dtype=np.float64
)

test_expected_severity = (
    test_tta_probs
    * severity_values
).sum(axis=1)

print(
    "Expected severity shape:",
    test_expected_severity.shape
)

print(
    "First 10 values:"
)

print(
    test_expected_severity[:10]
)
```

```python
def apply_thresholds(
    severity,
    thresholds
):
    return np.digitize(
        severity,
        np.asarray(
            thresholds,
            dtype=np.float64
        )
    )


test_threshold_predictions = (
    apply_thresholds(
        test_expected_severity,
        best_thresholds
    )
)

print(
    "Threshold predictions:",
    test_threshold_predictions.shape
)

print(
    "Predicted classes:",
    np.unique(
        test_threshold_predictions
    )
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    cohen_kappa_score,
    matthews_corrcoef,
    log_loss,
    roc_auc_score
)

def evaluate_predictions(
    y_true,
    probs,
    predictions
):

    y_true = np.asarray(y_true)
    probs = np.asarray(
        probs,
        dtype=np.float64
    )
    predictions = np.asarray(
        predictions
    )

    probs = (
        probs
        / np.clip(
            probs.sum(
                axis=1,
                keepdims=True
            ),
            1e-12,
            None
        )
    )

    results = {}

    results["Accuracy"] = accuracy_score(
        y_true,
        predictions
    )

    results["Balanced Accuracy"] = (
        balanced_accuracy_score(
            y_true,
            predictions
        )
    )

    results["Macro Precision"] = (
        precision_score(
            y_true,
            predictions,
            average="macro",
            zero_division=0
        )
    )

    results["Macro Recall"] = (
        recall_score(
            y_true,
            predictions,
            average="macro",
            zero_division=0
        )
    )

    results["Macro F1"] = (
        f1_score(
            y_true,
            predictions,
            average="macro",
            zero_division=0
        )
    )

    results["Weighted F1"] = (
        f1_score(
            y_true,
            predictions,
            average="weighted",
            zero_division=0
        )
    )

    results["QWK"] = (
        cohen_kappa_score(
            y_true,
            predictions,
            weights="quadratic"
        )
    )

    results["Cohen Kappa"] = (
        cohen_kappa_score(
            y_true,
            predictions
        )
    )

    results["MCC"] = (
        matthews_corrcoef(
            y_true,
            predictions
        )
    )

    results["Log Loss"] = log_loss(
        y_true,
        probs,
        labels=np.arange(5)
    )

    try:
        results["ROC-AUC"] = roc_auc_score(
            y_true,
            probs,
            multi_class="ovr",
            average="macro"
        )
    except Exception:
        results["ROC-AUC"] = np.nan

    return results
```

```python
test_tta_metrics = evaluate_predictions(
    test_tta_y_true,
    test_tta_probs,
    test_tta_predictions
)

test_threshold_metrics = evaluate_predictions(
    test_tta_y_true,
    test_tta_probs,
    test_threshold_predictions
)

test_comparison = pd.DataFrame({
    "224 Horizontal Flip TTA":
        test_tta_metrics,

    "TTA + Frozen Thresholds":
        test_threshold_metrics
})

display(test_comparison)
```
